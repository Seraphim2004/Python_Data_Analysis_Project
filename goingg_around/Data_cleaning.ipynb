```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.loc[:, 'salary_rate':'salary_hour_avg'].dropna(subset='salary_rate')
```

```python
median_yearly_sal = df['salary_year_avg'].median()
```

```python
median_hourly_sal = df['salary_hour_avg'].median()
```

```python
df_filled = df
df_filled['salary_year_avg']=df_filled['salary_year_avg'].fillna(median_yearly_sal)
df_filled['salary_hour_avg']=df_filled['salary_hour_avg'].fillna(median_hourly_sal)
```

```python
df_filled.loc[:,'salary_year_avg':'salary_hour_avg']
```

```python
df_unique=df_filled
df_unique = df_unique.drop_duplicates()

print("Length of original df = ",len(df_filled))
print("Length of duplicate rows dropped df = ",len(df_unique))
print("Total rows deleted = ",len(df_filled) - len(df_unique))
```

```python
df_unique = df_unique.drop_duplicates(subset = ['job_title','company_name'])

print("Length of original df = ",len(df_filled))
print("Length of duplicate rows dropped df = ",len(df_unique))
print("Total rows deleted = ",len(df_filled) - len(df_unique))
```
